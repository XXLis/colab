# Analyse van Huurprijzen Dataset

Deze notebook analyseert de verdeling van huurprijzen uit de House Rent Dataset. We maken gebruik van Seaborn om histograms te maken met verschillende aantallen bins.

## 1. Importeer benodigde bibliotheken

Eerst importeren we de bibliotheken die nodig zijn voor data-analyse en visualisatie.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2. Data laden en verkennen

We laden de dataset en bekijken de eerste rijen om een indruk te krijgen van de data. Dit helpt ons om de structuur van de dataset te begrijpen.

```python
# Laad de dataset
df = pd.read_csv('House_Rent_Dataset.csv')

# Toon de eerste 5 rijen
df.head()
```

## 3. Eerste histogram - standaard aantal bins

We maken een histogram met het standaard aantal bins om een algemeen overzicht te krijgen van de verdeling van huurprijzen.

```python
plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='Rent')
plt.title('Verdeling van huurprijzen - standaard aantal bins', fontsize=16)
plt.xlabel('Huurprijs (Rent)', fontsize=14)
plt.ylabel('Aantal advertenties', fontsize=14)
plt.show()
```

## 4. Tweede histogram - 50 bins

Nu maken we een histogram met 50 bins voor een gedetailleerdere weergave van de verdeling. Dit laat meer details zien in de data.

```python
plt.figure(figsize=(12, 8))
sns.histplot(data=df, x='Rent', bins=50)
plt.title('Verdeling van huurprijzen - 50 bins', fontsize=16)
plt.xlabel('Huurprijs (Rent)', fontsize=14)
plt.ylabel('Aantal advertenties', fontsize=14)
plt.show()
```
